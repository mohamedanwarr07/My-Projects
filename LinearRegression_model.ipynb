```python
df['Tutoring_per_StudyHour'] = (
    df['Tutoring_Sessions'] / df['Hours_Studied']
)
df['Study_Sleep_Ratio'] = df['Hours_Studied'] / df['Sleep_Hours']
```

```python
df.drop(columns=["Tutoring_per_StudyHour"],inplace=True)
```

```python
x=df.drop(columns=["Exam_Score"])
y=df["Exam_Score"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
num_fea=x_train.select_dtypes(include=["int64","float64"]).columns
cat_fea=x_train.select_dtypes(include=["object"]).columns
print("num",num_fea)
print("cat",cat_fea)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
num_transformer=Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
])
```

```python
from sklearn.preprocessing import OneHotEncoder

cat_transformer=Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])
```

```python
from sklearn.compose import ColumnTransformer
preprocessor=ColumnTransformer([
    ("num",num_transformer,num_fea),
    ("cat",cat_transformer,cat_fea)
])
```

```python
print(num_fea)
print(cat_fea)
```

```python
X_train_pre=preprocessor.fit_transform(x_train)
X_test_pre=preprocessor.transform(x_test)
```

```python
print(X_train_pre.shape)
print(X_test_pre.shape)
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
baseline=DummyRegressor()
baseline.fit(X_train_pre,y_train)
y_pred=baseline.predict(X_test_pre)
print("MAE",mean_absolute_error(y_test,y_pred))
print("MSE",mean_squared_error(y_test,y_pred))
print("R2",r2_score(y_test,y_pred))
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train_pre,y_train)
y_pred=model.predict(X_test_pre)
print("MAE",mean_absolute_error(y_test,y_pred))
print("MSE",mean_squared_error(y_test,y_pred))
print("R2",r2_score(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeRegressor
tree=DecisionTreeRegressor(random_state=42)
tree.fit(X_train_pre,y_train)
y_pred=tree.predict(X_test_pre)
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

```python
tree_pred=tree.predict(X_train_pre)
test_predict=tree.predict(X_test_pre)
print(mean_absolute_error(y_train,tree_pred))
print(mean_absolute_error(y_test,test_predict))
print(r2_score(y_train,tree_pred))
print(r2_score(y_test,test_predict))
```

```python
import numpy as np
tree = DecisionTreeRegressor(
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

tree.fit(X_train_pre, y_train)

tree_pred = tree.predict(X_test_pre)

print("Train R2:", r2_score(y_train, tree.predict(X_train_pre)))
print("Test R2:", r2_score(y_test, tree_pred))
print("MAE:", mean_absolute_error(y_test, tree_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, tree_pred)))
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(n_estimators=100,
                         random_state=42,
                         n_jobs=-1)
rf.fit(X_train_pre,y_train)
y_pred=rf.predict(X_test_pre)
print("mae",mean_absolute_error(y_test,y_pred))
print("mse",mean_squared_error(y_test,y_pred))
print("r2",r2_score(y_test,y_pred))
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gb.fit(X_train_pre, y_train)

gb_pred = gb.predict(X_test_pre)

print("MAE:", mean_absolute_error(y_test, gb_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, gb_pred)))
print("R2:", r2_score(y_test, gb_pred))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid={
    "n_estimators":[100,200,300],
    "learning_rate":[0.01,0.05,0.1],
    "max_depth":[2,3,4]
}
grid = GridSearchCV(
    GradientBoostingRegressor(random_state=42),
    param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    error_score='raise'
)

grid.fit(X_train_pre, y_train)
print(grid.best_params_)
print(grid.best_score_)
```

```python
best_model = grid.best_estimator_
tuned_pred=best_model.predict(X_test_pre)
print("MAE:", mean_absolute_error(y_test, tuned_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, tuned_pred)))
print("R2:", r2_score(y_test, tuned_pred))
```
